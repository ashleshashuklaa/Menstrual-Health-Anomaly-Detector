```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = '/content/drive/MyDrive/SheCares/dhs/IAIR7EFL.DTA'

df = pd.read_stata(file_path,
                   columns=["v012", "v024", "v025", "v215"],
                   convert_categoricals=True)

# Figure 1 — Age distribution
plt.figure(figsize=(10, 5))
df["v012"].hist(bins=35, color="mediumpurple")
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution of Indian Women (NFHS-5)")
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/SheCares/figures /nfhs_age_distribution.png", dpi=150)
plt.close()
print("Figure 1 done")

# Figure 2 — Rural vs Urban
plt.figure(figsize=(6, 5))
df["v025"].value_counts().plot(kind="bar", color=["steelblue", "coral"])
plt.title("Rural vs Urban Split (NFHS-5)")
plt.ylabel("Number of Women")
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/SheCares/figures /nfhs_rural_urban.png", dpi=150)
plt.close()
print("Figure 2 done")

# Figure 3 — Top 10 states
plt.figure(figsize=(12, 5))
df["v024"].value_counts().head(10).plot(kind="bar", color="teal")
plt.title("Top 10 States Represented (NFHS-5)")
plt.ylabel("Number of Women")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/SheCares/figures /nfhs_top_states.png", dpi=150)
plt.close()
print("Figure 3 done")

# Stats for paper
print("\nTotal women surveyed:", len(df))
print("\nAge stats:")
print(df["v012"].describe())
print("\nResidence split (%):")
print(df["v025"].value_counts(normalize=True) * 100)
print("\nTop 5 states:")
print(df["v024"].value_counts().head(5))
```

```python
import pandas as pd

file_path = '/content/drive/MyDrive/SheCares/dhs/IAIR7EFL.DTA'

# Load only first 50000 rows to test
df = pd.read_stata(file_path,
                   columns=["v012", "v024", "v025"],
                   convert_categoricals=True,
                   chunksize=50000)

# Take only first chunk
df = next(iter(df))
print("Loaded successfully!")
print("Shape:", df.shape)
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Figure 1 — Age distribution
plt.figure(figsize=(10, 5))
df["v012"].hist(bins=35, color="mediumpurple")
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution of Indian Women (NFHS-5)")
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/SheCares/figures /nfhs_age_distribution.png", dpi=150)
plt.close()
print("Figure 1 done ✅")

# Figure 2 — Rural vs Urban
plt.figure(figsize=(6, 5))
df["v025"].value_counts().plot(kind="bar", color=["steelblue", "coral"])
plt.title("Rural vs Urban Split (NFHS-5)")
plt.ylabel("Number of Women")
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/SheCares/figures /nfhs_rural_urban.png", dpi=150)
plt.close()
print("Figure 2 done ✅")

# Figure 3 — Top 10 states
plt.figure(figsize=(12, 5))
df["v024"].value_counts().head(10).plot(kind="bar", color="teal")
plt.title("Top 10 States Represented (NFHS-5)")
plt.ylabel("Number of Women")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/SheCares/figures /nfhs_top_states.png", dpi=150)
plt.close()
print("Figure 3 done ✅")

# Stats for paper
print("\nTotal rows used:", len(df))
print("\nAge stats:")
print(df["v012"].describe())
print("\nResidence split (%):")
print(df["v025"].value_counts(normalize=True) * 100)
print("\nTop 5 states:")
print(df["v024"].value_counts().head(5))
```
